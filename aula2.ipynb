## 1. Revisão — crescimento econômico

Execute este exemplo e identifique variáveis, operadores e tipos de dados.
Esta aula pode ser executada independentemente do arquivo da Aula 1.


````python
pib_2025 = 5000
pib_2026 = 5148
crescimento = (pib_2026 / pib_2025 - 1) * 100

print("Crescimento do PIB (%):", crescimento)
````

### Strings e f-strings

Texto entre aspas é uma string. Uma **f-string** começa com `f` e insere os valores
das expressões entre `{ }`. `:.2f` mostra um número com duas casas decimais.
Isso muda a apresentação, sem alterar o valor guardado na variável.


````python
pais = "Brasil"
ano = 2026

crescimento_int = int(crescimento) # Transforma crescimento em "int". Exclui a parte após o "." (não é um arredondamento)

crescimento_round = round(crescimento,2) # Arredonda para 2 casas decimais

print(f"O crescimento do PIB do {pais} em {ano} foi de {crescimento:.2f}%.")

print("O crescimento do PIB do ",pais,"em",ano,"foi de",crescimento_int,"%")

print("O crescimento do PIB do ",pais,"em",ano,"foi de",crescimento_round,"%")
````

## 2. Listas — guardar vários valores em sequência

Podemos distinguir, os tipos que representam um único valor daqueles que permitem agrupar vários valores.

| Categoria | Tipo em Python | Exemplo | O que representa |
|---|---|---|---|
| Simples | `int` | `ano = 2024` | Um número inteiro |
| Simples | `float` | `inflacao = 4.2` | Um número com casas decimais |
| Simples | `bool` | `acima_da_meta = True` | Um valor lógico: verdadeiro ou falso |
| Composto | `list` | `inflacoes = [6.0, 4.5, 4.2]` | Uma sequência de valores, acessados pela posição |
| Composto | `dict` | `dados = {"ano": 2024, "inflacao": 4.2}` | Um conjunto de pares de chave e valor |


Uma lista ou um dicionário pode conter valores de diferentes tipos, inclusive outras listas e outros dicionários.
E os textos? O tipo str, como em pais = "Brasil", é uma sequência de caracteres. Embora muitas vezes seja apresentado junto dos tipos básicos, ele permite acessar caracteres pela posição: pais[0] retorna "B".





Uma lista usa colchetes e separa seus elementos por vírgulas.
Aqui, os valores de inflação estão em **porcentagem**: `6.0` significa 6%.

````python
anos = [2022, 2023, 2024, 2025, 2026]
inflacao_lista = [6.0, 4.5, 4.2, 5.0, 3.5]

print(anos)
print(inflacao_lista)
print(type(inflacao_lista))
````

### Índices — Python começa a contar em zero

| Posição | Ano | Inflação (%) |
|---|---|---|
| `0` | 2022 | 6.0 |
| `1` | 2023 | 4.5 |
| `2` | 2024 | 4.2 |
| `3` | 2025 | 5.0 |
| `4` | 2026 | 3.5 |

`lista[0]` pega o primeiro elemento; `lista[-1]` pega o último.
Nessas duas listas, o ano e sua inflação estão na mesma posição.


````python
print("Primeiro ano:", anos[0])
print("Inflação do primeiro ano:", inflacao_lista[0])
print("Inflação em 2024:", inflacao_lista[2])
print("Última inflação:", inflacao_lista[-1])
````

### Funções - introdução rápida


Uma função é um conjunto de instruções que realiza uma tarefa. Ela pode receber valores de entrada, chamados argumentos, e devolver um resultado.

O Python já vem com várias funções embutidas, que podemos usar diretamente, sem precisar importar módulos ou criá-las.

| Função | O que faz | Exemplo |
|---|---|---|
| `print()` | Exibe informações | `print("Olá!")` |
| `type()` | Informa o tipo de um objeto | `type(4.2)` |
| `len()` | Conta os elementos | `len([6.0, 4.5, 4.2])` |
| `sum()` | Calcula a soma | `sum([6.0, 4.5, 4.2])` |
| `min()` | Retorna o menor valor | `min([6.0, 4.5, 4.2])` |
| `max()` | Retorna o maior valor | `max([6.0, 4.5, 4.2])` |
| `round()` | Arredonda um número | `round(4.237, 2)` |

### Quantidade de observações e média

`len()` conta os elementos e `sum()` soma valores numéricos.
Calculamos abaixo a média aritmética das taxas anuais; ela não é a inflação acumulada.


````python
quantidade_anos = len(inflacao_lista)
inflacao_media = sum(inflacao_lista) / quantidade_anos

print("Quantidade de anos:", quantidade_anos)
print(f"Inflação média: {inflacao_media:.2f}%")
````

## 4. Dicionários — acessar pelo significado

Na lista, `inflacao_lista[2]` acessa a terceira posição.
Em um dicionário, podemos associar uma **chave** (o ano) a um **valor** (a inflação).

Criamos dicionários com `{ }`, escrevemos `chave: valor` e consultamos com `[chave]`.


````python
inflacao_por_ano = {
    2022: 6.0,
    2023: 4.5,
    2024: 4.2,
    2025: 5.0,
    2026: 3.5
}

print(inflacao_por_ano[2024])
print(type(inflacao_por_ano))
````

### Adicionar e alterar

A mesma forma de atribuição serve para adicionar uma chave nova ou mudar seu valor.
A célula abaixo pode ser reexecutada: a chave 2027 continuará aparecendo uma só vez.


````python
inflacao_por_ano[2027] = 3.2
print("Valor adicionado:", inflacao_por_ano[2027])

inflacao_por_ano[2027] = 3.1
print("Valor alterado:", inflacao_por_ano[2027])
````

## 5. Uma base única — dicionário por ano e por variável

Podemos guardar um dicionário dentro de outro. Essa estrutura é chamada de dicionário aninhado. Cada ano reúne três variáveis:

- `pib`: PIB real, a preços constantes, na mesma unidade em todos os anos.
- `inflacao`: inflação anual, em %.
- `desemprego`: taxa de desemprego, em %.

Primeiro escolhemos o **ano**, depois a **variável**.


````python
dados = {
    2022: {"pib": 1000, "inflacao": 6.0, "desemprego": 8.2},
    2023: {"pib": 1020, "inflacao": 4.5, "desemprego": 7.8},
    2024: {"pib": 1050, "inflacao": 4.2, "desemprego": 7.1},
    2025: {"pib": 1040, "inflacao": 5.0, "desemprego": 6.7},
    2026: {"pib": 1080, "inflacao": 3.5, "desemprego": 6.5}
}

dados
````

### Passo 1 — escolher um ano

`dados[2024]` devolve o dicionário com as variáveis de 2024.


````python
dados[2024]
````

### Passo 2 — escolher a variável

`dados[2024]["inflacao"]` acessa a inflação dentro do dicionário daquele ano.
As chaves dos anos são inteiros; as chaves das variáveis são strings.


````python
print("Inflação de 2024:", dados[2024]["inflacao"])
print("PIB de 2024:", dados[2024]["pib"])
````

### Se os dois colchetes parecerem confusos

Podemos guardar o dicionário de um ano em uma variável intermediária e depois consultar a inflação.
O resultado é o mesmo.


````python
dados_2024 = dados[2024]
inflacao_2024 = dados_2024["inflacao"]
print(inflacao_2024)
````

### Aplicação — crescimento do PIB

Usamos a mesma equação da Aula 1, agora consultando os valores na base.
Como o PIB já está a preços constantes, não subtraímos a inflação deste crescimento.


````python
crescimento_2026 = (dados[2026]["pib"] / dados[2025]["pib"] - 1) * 100
print(f"Crescimento real do PIB em 2026: {crescimento_2026:.2f}%")
````

### Aplicação — inflação acima da meta?

Uma variável pode guardar o resultado de uma comparação.


````python
meta_inflacao = 3.0
acima_da_meta = dados[2026]["inflacao"] > meta_inflacao

print("Inflação acima da meta?", acima_da_meta)
````
