## 2. Condicionais — escolher o que executar
### `if`: executar um bloco quando a condição é verdadeira

Leia `if` como **“se”**. A linha da condição termina em `:`. O bloco abaixo começa com **quatro espaços**.
Se a condição for falsa, o Python pula esse bloco e continua depois dele.

*Referência: Think Python, seção [5.4](https://allendowney.github.io/ThinkPython/chap05.html#conditional-execution).*


````python
inflacao = 4.2

if inflacao > meta_inflacao:
    print("Inflação acima da meta.")

print("Consulta concluída.")
````

**Experimente:** altere `inflacao` para `2.8` e execute novamente. Qual mensagem desaparece?
O último `print` está fora do bloco: ele será executado em ambos os casos.

### `else`: executar uma alternativa

Leia `else` como **“caso contrário”**. Um `if` com `else` executa exatamente um dos dois blocos.
O `else` fica alinhado com o `if`, termina em `:` e não recebe uma condição.

*Referência: Think Python, seção [5.5](https://allendowney.github.io/ThinkPython/chap05.html#the-else-clause).*


````python
inflacao = 2.8

if inflacao > meta_inflacao:
    print("Inflação acima da meta.")
else:
    print("Inflação igual ou inferior à meta.")
````

### `elif`: distinguir três possibilidades

Leia `elif` como **“senão, se”**. As condições são verificadas de cima para baixo.
Ao encontrar a primeira verdadeira, o Python executa aquele bloco e encerra a seleção.

Aqui, queremos distinguir **acima**, **abaixo** e **igual à meta**.
O `else` cobre a igualdade porque as duas condições anteriores já eliminaram as outras possibilidades.

*Referência: Think Python, seção [5.6](https://allendowney.github.io/ThinkPython/chap05.html#chained-conditionals).*


````python
inflacao = 3.0

if inflacao > meta_inflacao:
    situacao = "acima da meta"
elif inflacao < meta_inflacao:
    situacao = "abaixo da meta"
else:
    situacao = "igual à meta"

print(f"Inflação de {inflacao:.1f}%: {situacao}.")
````

**Experimente:** execute a célula com `inflacao = 4.2`, `2.8` e `3.0`. Os três caminhos devem funcionar.
Nesta atividade, os números são escolhidos diretamente. Em cálculos com números decimais, pequenas diferenças de representação podem exigir uma comparação com tolerância; retomaremos isso em outra aula.

### Atenção à indentação

| Linha | Papel |
|---|---|
| `if condicao:` | Abre a seleção |
| Quatro espaços antes de uma instrução | Coloca a instrução dentro do bloco |
| `elif outra_condicao:` / `else:` | Abre uma alternativa, no mesmo alinhamento do `if` |
| Voltar ao alinhamento inicial | Continua depois da seleção |

## 3. Combinar condições

| Operador | Significado |
|---|---|
| `and` | Verdadeiro quando as duas condições são verdadeiras |
| `or` | Verdadeiro quando pelo menos uma condição é verdadeira |
| `not` | Inverte um resultado lógico |

Vamos identificar uma observação com inflação acima de 3% **e** desemprego acima de 7%.
Esses limites são critérios didáticos de seleção, não uma recomendação de política econômica.

*Referência: Think Python, seção [5.3](https://allendowney.github.io/ThinkPython/chap05.html#logical-operators).*


````python
ano_consulta = 2024
inflacao = dados[ano_consulta]["inflacao"]
desemprego = dados[ano_consulta]["desemprego"]

print("As duas condições:", (inflacao > 3.0) and (desemprego > 7.0))
print("Pelo menos uma:", (inflacao > 3.0) or (desemprego > 7.0))
print("Inflação não está acima:", not (inflacao > 3.0))
````

## 4. Repetição — executar o mesmo bloco para cada elemento

Um **laço de repetição** executa um bloco várias vezes. Cada passagem é uma **iteração**.
Com `for`, a variável do laço recebe, em sequência, cada elemento da lista.
Ela recebe o **valor do elemento**, não sua posição.

*Referência: Think Python, seção [9.7](https://allendowney.github.io/ThinkPython/chap09.html#looping-through-a-list).*


````python
taxas = [6.0, 4.5, 4.2]

for taxa in taxas:
    print(f"Inflação: {taxa:.1f}%")

print("Fim da lista.")
````

### Acompanhar a execução

| Iteração | Valor de `taxa` | Ação |
|---|---:|---|
| 1 | 6.0 | Exibir 6,0% |
| 2 | 4.5 | Exibir 4,5% |
| 3 | 4.2 | Exibir 4,2% |

`taxa` é apenas um nome escolhido por nós; poderia ser outro. O `print` indentado roda três vezes.
O `print` fora do bloco roda uma vez, depois do laço. Se a lista for vazia, o bloco do `for` não será executado.

### Colocar uma decisão dentro da repetição

Há dois níveis de indentação: quatro espaços para o bloco do `for`; oito para o bloco do `if`/`else` dentro dele.


````python
taxas_teste = [4.2, 3.0, 2.8]

for taxa in taxas_teste:
    if taxa > meta_inflacao:
        print(f"{taxa:.1f}%: acima da meta")
    else:
        print(f"{taxa:.1f}%: igual ou inferior à meta")
````

## 5. Percorrer a base econômica

Quando escrevemos `for ano in dados:`, percorremos as **chaves** do dicionário.
Neste caso, cada chave é um ano. `dados[ano]` acessa o registro correspondente.

O dicionário percorre as chaves na ordem de inserção. Não ordena os anos automaticamente.
A base deste notebook foi criada com os anos em ordem cronológica.

*Referência: Think Python, seção [10.5](https://allendowney.github.io/ThinkPython/chap10.html#looping-and-dictionaries).*


````python
for ano in dados:
    inflacao_ano = dados[ano]["inflacao"]
    desemprego_ano = dados[ano]["desemprego"]
    print(f"{ano}: inflação {inflacao_ano:.1f}% | desemprego {desemprego_ano:.1f}%")
````

### Filtrar na exibição

A condição dentro do laço permite exibir apenas os anos que atendem a um critério.
Este exemplo imprime somente observações com desemprego inferior a 7%; a base original continua completa.


````python
for ano in dados:
    desemprego_ano = dados[ano]["desemprego"]
    if desemprego_ano < 7.0:
        print(f"{ano}: desemprego de {desemprego_ano:.1f}%")
````

## 6. Repetir cálculos entre anos

Para calcular crescimento precisamos do PIB de dois anos. A base começa em 2022, mas não contém 2021.
Portanto, o primeiro crescimento que podemos calcular é o de **2023 em relação a 2022**.

`anos[1:]` seleciona a lista a partir do segundo elemento. Nesta base, os anos são consecutivos:
`ano - 1` identifica o ano anterior. Se houvesse lacunas, essa estratégia precisaria ser ajustada.


````python
print("Anos com crescimento calculável:", anos[1:])

for ano in anos[1:]:
    pib_atual = dados[ano]["pib"]
    pib_anterior = dados[ano - 1]["pib"]
    crescimento_ano = (pib_atual / pib_anterior - 1) * 100
    print(f"{ano}: crescimento real de {crescimento_ano:.2f}%")
````

### Guardar resultados em vez de apenas exibi-los

Criamos um dicionário vazio **antes** do laço. A cada iteração, acrescentamos o crescimento daquele ano.
Se criássemos `{}` dentro do laço, apagaríamos os resultados anteriores a cada passagem.
Ao reexecutar esta célula, começamos novamente com o dicionário vazio.


````python
crescimentos = {}

for ano in anos[1:]:
    pib_atual = dados[ano]["pib"]
    pib_anterior = dados[ano - 1]["pib"]
    crescimentos[ano] = (pib_atual / pib_anterior - 1) * 100

print(crescimentos)
print(f"Crescimento de 2025: {crescimentos[2025]:.2f}%")
````

**Interpretação:** 2025 tem crescimento anual negativo; os demais crescimentos calculados são positivos.
Como o PIB já está a preços constantes, essas variações são reais: não subtraímos a inflação.
Os números guardados não são arredondados. `:.2f` altera somente sua apresentação.

## 8. Extensão — contar e guardar os anos selecionados (para casa)

Um **contador** começa em zero e aumenta a cada ocorrência de interesse.
Uma lista vazia pode receber os anos selecionados com `.append()`.
O contador e a lista precisam ser inicializados antes do laço.

*Referências: Think Python, seções [7.3–7.4](https://allendowney.github.io/ThinkPython/chap07.html#updating-variables) e [10.7](https://allendowney.github.io/ThinkPython/chap10.html#accumulating-a-list). Leitura complementar: concentre-se na atualização de variáveis e na acumulação; os exemplos de funções do livro serão retomados na aula 4.*


````python
quantidade = 0
anos_selecionados = []

for ano in dados:
    if dados[ano]["desemprego"] < 7.0:
        quantidade = quantidade + 1
        anos_selecionados.append(ano)

print("Quantidade:", quantidade)
print("Anos:", anos_selecionados)
````
